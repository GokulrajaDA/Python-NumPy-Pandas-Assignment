```python
import numpy as np
import pandas as pd

# PART A: NUMPY ARRAY OPERATIONS - Temperature Analysis

print("========== PART A: NUMPY ==========")

# 1. Create 1D NumPy Array
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("\n1. Week 1 Temperatures Array:")
print(temperatures_w1)
```

```python
# 2. Inspection and Properties
print("\n2. Inspection:")
print(f"Shape: {temperatures_w1.shape}")
print(f"Data Type (dtype): {temperatures_w1.dtype}")
print(f"Number of elements (size): {temperatures_w1.size}")
print(f"Dimensions (ndim): {temperatures_w1.ndim}")
```

```python
# 3. Array Operations
# Celsius to Fahrenheit
temperatures_f = (temperatures_w1 * 9/5) + 32
print("\n3. Array Operations:")
print(f"Temperatures in Fahrenheit: {temperatures_f}")
print(f"Max Temp (C): {np.max(temperatures_w1)}")
print(f"Min Temp (C): {np.min(temperatures_w1)}")
print(f"Mean Temp (C): {np.mean(temperatures_w1):.2f}")
print(f"Mean Temp (F): {np.mean(temperatures_f):.2f}")
```

```python
# 4. Array Slicing and Indexing
print("\n4. Slicing & Indexing:")
print(f"First three days: {temperatures_w1[0:3]}") # or [:3]
print(f"Weekend (last two days): {temperatures_w1[-2:]}")
print(f"Middle three days (Tue-Thu): {temperatures_w1[2:5]}") # index 2,3,4
```

```python
# 5. Create a 2D Array
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9], # Week 1
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1] # Week 2
])
print("\n5. 2D Array (2 Weeks):")
print(temperatures)
```

```python
# 6. Inspect and Slice the 2D Array
print("\n6. 2D Array Inspection & Slicing:")
print(f"Shape: {temperatures.shape} -> 2 rows (weeks), 7 cols (days)")
print(f"Dtype: {temperatures.dtype}")
print(f"Total elements: {temperatures.size}")
print(f"Week 1: {temperatures[0]}")
print(f"Week 2: {temperatures[1]}")
print(f"Weekends for both weeks (last 2 days): \n{temperatures[:, -2:]}")
print(f"Week 1 Weekend: {temperatures[0, -2:]}")
print(f"Week 2 Weekend: {temperatures[1, -2:]}")
```

```python
# PART B: PANDAS SERIES

print("\n\n========== PART B: PANDAS SERIES ==========")

# 1. Creating Pandas Series
marks = pd.Series([95, 92, 89, 85, 80], index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'], name='Marks')
print("\n1. Series 'marks':")
print(marks)
```

```python
# 2. Indexing and Slicing
print("\n2. Indexing & Slicing:")
print(f"1st rank student mark (integer position [0]): {marks.iloc[0]}")
# Using integer index position
print(f"1st rank mark using [0]: {marks[0]}")

print(f"\nTop 3 ranks using loc:\n{marks.loc['Rank1':'Rank3']}")
print(f"\n3rd-ranked student using iloc[2]: {marks.iloc[2]}")
print(f"\nMarks > 90 (Boolean Mask):\n{marks[marks > 90]}")
```

```python
# 3. Manipulating Series
print("\n3. Manipulating Series:")
marks['Rank1'] = 100
print(f"After modifying Rank1 to 100:\n{marks}")

marks = marks.drop('Rank5')
print(f"\nAfter removing last rank (Rank5):\n{marks}")

cgpa = marks / 10
print(f"\nCGPA (Marks/10):\n{cgpa}")
```

```python
# PART C: PANDAS DATAFRAME

print("\n\n========== PART C: PANDAS DATAFRAME ==========")

# 1. Creating DataFrame
data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}
transactions = pd.DataFrame(data)
print("\n1. Transactions DataFrame:")
print(transactions)
```

```python
# 2. Data Exploration
print("\n2. Data Exploration:")
print("\nHead (first 5):")
print(transactions.head())
print("\nTail (last 5):")
print(transactions.tail())
print(f"\nShape: {transactions.shape}")
print(f"Column Names: {list(transactions.columns)}")
print(f"Data Types:\n{transactions.dtypes}")
print(f"\nInfo:")
print(transactions.info())

print("\nOnly ProductCategory and Amount columns:")
print(transactions[['ProductCategory', 'Amount']])

print("\nLast 3 columns using iloc:")
print(transactions.iloc[:, -3:]) # last 3 cols = ProductCategory, Region, Amount

print("\nFilter: Region='North' and Amount>200:")
filtered = transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]
print(filtered)

print("\nValue counts for ProductCategory:")
print(transactions['ProductCategory'].value_counts())

print("\nUnique values in Region:")
print(transactions['Region'].unique())

print("\nGroup by Region and mean Amount:")
print(transactions.groupby('Region')['Amount'].mean())
```

```python
# 3. Manipulating the DataFrame
print("\n3. Manipulating DataFrame:")

# Modify Amount for TransactionID 102 to 165
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print("\nAfter modifying TransactionID 102 Amount to 165:")
print(transactions)

# Add new column Discount (10% of Amount)
transactions['Discount'] = transactions['Amount'] * 0.10
print("\nAfter adding Discount column (10%):")
print(transactions)

# Remove row with TransactionID 109
transactions = transactions[transactions['TransactionID']!= 109]
print("\nAfter removing TransactionID 109:")
print(transactions)

# Delete Discount column
transactions = transactions.drop(columns=['Discount'])
print("\nAfter deleting Discount column (Final DataFrame):")
print(transactions)
```
